```python
events["datetime"] = pd.to_datetime(
    events["timestamp"], unit="ms", utc=True
)

print("Строк:", len(events))
print("Пользователей:", events["visitorid"].nunique())
print("Товаров:", events["itemid"].nunique())
print("Начало:", events["datetime"].min())
print("Конец:", events["datetime"].max())
print("Полных дубликатов:", events.duplicated().sum())

history_lengths = events.groupby("visitorid").size()

display(
    history_lengths.describe(percentiles=[0.5, 0.9, 0.99])
)

print(
    "Доля пользователей с единственным событием:",
    f"{history_lengths.eq(1).mean():.1%}",
)
```

```python
from pathlib import Path
import pandas as pd
from IPython.display import display

data_dir = Path("data/raw/dunnhumby")

transactions = pd.read_csv(data_dir / "transaction_data.csv")
products = pd.read_csv(data_dir / "product.csv")
households = pd.read_csv(data_dir / "hh_demographic.csv")

display(transactions.head())
display(products.head())
display(households.head())

print("Строк покупок:", len(transactions))
print("Домохозяйств:", transactions["household_key"].nunique())
print("Товаров в покупках:", transactions["PRODUCT_ID"].nunique())
print("Корзин:", transactions["BASKET_ID"].nunique())
print("Диапазон DAY:",
      transactions["DAY"].min(), transactions["DAY"].max())

basket_counts = (
    transactions.groupby("household_key")["BASKET_ID"].nunique()
)

display(basket_counts.describe(percentiles=[0.5, 0.9, 0.99]))

print(
    "Доля домохозяйств с одной корзиной:",
    f"{basket_counts.eq(1).mean():.1%}"
)

print("Строк с QUANTITY ≤ 0:",
      transactions["QUANTITY"].le(0).sum())
display(transactions.isna().sum())
```
