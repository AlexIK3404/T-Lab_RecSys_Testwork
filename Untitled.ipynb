```python
from pathlib import Path
import pandas as pd
from IPython.display import display

data_dir = Path("data/raw/dunnhumby")

transactions = pd.read_csv(data_dir / "transaction_data.csv")
products = pd.read_csv(data_dir / "product.csv")
households = pd.read_csv(data_dir / "hh_demographic.csv")

display(transactions.head())
display(products.head())
display(households.head())

print("Строк покупок:", len(transactions))
print("Домохозяйств:", transactions["household_key"].nunique())
print("Товаров в покупках:", transactions["PRODUCT_ID"].nunique())
print("Корзин:", transactions["BASKET_ID"].nunique())
print("Диапазон DAY:",
      transactions["DAY"].min(), transactions["DAY"].max())

basket_counts = (
    transactions.groupby("household_key")["BASKET_ID"].nunique()
)

display(basket_counts.describe(percentiles=[0.5, 0.9, 0.99]))

print(
    "Доля домохозяйств с одной корзиной:",
    f"{basket_counts.eq(1).mean():.1%}"
)

print("Строк с QUANTITY ≤ 0:",
      transactions["QUANTITY"].le(0).sum())
display(transactions.isna().sum())
```

```python
display(pd.Series({
    "QUANTITY == 0": transactions["QUANTITY"].eq(0).sum(),
    "QUANTITY < 0": transactions["QUANTITY"].lt(0).sum(),
}))

# Для первой версии покупкой считаем строку с QUANTITY > 0.
# Исходную таблицу transactions сохраняем без изменений.
purchases = transactions.loc[
    transactions["QUANTITY"] > 0
].copy()

product_features = products[
    [
        "PRODUCT_ID",
        "DEPARTMENT",
        "COMMODITY_DESC",
        "SUB_COMMODITY_DESC",
        "BRAND",
    ]
]

purchases = purchases.merge(
    product_features,
    on="PRODUCT_ID",
    how="left",
    validate="many_to_one",
    indicator=True,
)

print("Строк положительных покупок:", len(purchases))
print(
    "Строк без соответствующего товара в справочнике:",
    purchases["_merge"].eq("left_only").sum(),
)

display(purchases[
    ["DEPARTMENT", "COMMODITY_DESC", "SUB_COMMODITY_DESC", "BRAND"]
].isna().sum())

purchases = purchases.drop(columns="_merge")

print(
    "Полных дубликатов в исходной таблице:",
    transactions.duplicated().sum(),
)
```

```python
HORIZON = 7

cutoffs = {
    "train": list(range(367, 564, 28)),
    "validation": [591, 619],
    "test": [647, 675, 703],
}

assert max(cutoffs["train"]) + HORIZON <= min(cutoffs["validation"])
assert max(cutoffs["validation"]) + HORIZON <= min(cutoffs["test"])
assert max(cutoffs["test"]) + HORIZON - 1 <= purchases["DAY"].max()
```

```python
t = cutoffs["train"][-1]  # 563

history = purchases.loc[purchases["DAY"] < t].copy()

recent_history = history.loc[history["DAY"] >= t - 90]
eligible_households = recent_history["household_key"].unique()

# Один положительный ответ на пару домохозяйство–товар,
# независимо от числа повторных покупок в целевую неделю.
positives = (
    purchases.loc[
        purchases["DAY"].ge(t)
        & purchases["DAY"].lt(t + HORIZON)
        & purchases["household_key"].isin(eligible_households),
        ["household_key", "PRODUCT_ID"],
    ]
    .drop_duplicates()
    .assign(label=1)
)

print("Момент рекомендации:", t)
print("Последний день истории:", history["DAY"].max())
print("Целевые дни:", t, "—", t + HORIZON - 1)
print("Домохозяйств для рекомендаций:", len(eligible_households))
print("Положительных пар:", len(positives))
print(
    "Домохозяйств с покупками в целевую неделю:",
    positives["household_key"].nunique(),
)
```

```python
# 1. Личные кандидаты.
personal_history = history.loc[
    history["DAY"].ge(t - 180)
    & history["household_key"].isin(eligible_households)
]

personal_counts = (
    personal_history.groupby(["household_key", "PRODUCT_ID"])["BASKET_ID"]
    .nunique()
    .rename("n_baskets")
    .reset_index()
)

personal_candidates = (
    personal_counts
    .sort_values(
        ["household_key", "n_baskets", "PRODUCT_ID"],
        ascending=[True, False, True],
    )
    .groupby("household_key", sort=False)
    .head(50)
    [["household_key", "PRODUCT_ID"]]
)

# 2. Общие популярные товары.
popular_products = (
    history.loc[history["DAY"].ge(t - 28)]
    .groupby("PRODUCT_ID")["BASKET_ID"]
    .nunique()
    .rename("n_baskets")
    .reset_index()
    .sort_values(
        ["n_baskets", "PRODUCT_ID"],
        ascending=[False, True],
    )
    .head(100)
    [["PRODUCT_ID"]]
)

popular_candidates = pd.DataFrame({
    "household_key": eligible_households
}).merge(popular_products, how="cross")

# 3. Объединяем и убираем пересечение двух источников.
candidates = (
    pd.concat(
        [personal_candidates, popular_candidates],
        ignore_index=True,
    )
    .drop_duplicates(["household_key", "PRODUCT_ID"])
    .reset_index(drop=True)
)

# 4. Только теперь присоединяем ответы.
candidates = candidates.merge(
    positives,
    on=["household_key", "PRODUCT_ID"],
    how="left",
    validate="one_to_one",
)
candidates["label"] = candidates["label"].fillna(0).astype("int8")

# 5. Проверяем покрытие будущих покупок.
future_counts = positives.groupby("household_key").size()

covered_counts = (
    candidates.loc[candidates["label"].eq(1)]
    .groupby("household_key")
    .size()
    .reindex(future_counts.index, fill_value=0)
)

print("Пар-кандидатов:", len(candidates))
print("Положительных среди кандидатов:", candidates["label"].sum())
print(
    "Доля положительных меток:",
    f"{candidates['label'].mean():.1%}",
)
print(
    "Покрытие всех положительных пар:",
    f"{covered_counts.sum() / future_counts.sum():.1%}",
)
print(
    "Среднее покрытие по домохозяйствам с покупками:",
    f"{(covered_counts / future_counts).mean():.1%}",
)
print(
    "Домохозяйств с покупками, но без попаданий в кандидаты:",
    covered_counts.eq(0).sum(),
)
```

```python
# Рейтинг личных товаров — по уже рассчитанной прошлой истории.
personal_ranked = (
    personal_counts
    .sort_values(
        ["household_key", "n_baskets", "PRODUCT_ID"],
        ascending=[True, False, True],
    )
    .copy()
)
personal_ranked["personal_rank"] = (
    personal_ranked.groupby("household_key").cumcount() + 1
)

# Полный рейтинг популярных товаров за прошлые 28 дней.
global_ranked = (
    history.loc[history["DAY"].ge(t - 28)]
    .groupby("PRODUCT_ID")["BASKET_ID"]
    .nunique()
    .rename("n_baskets")
    .reset_index()
    .sort_values(
        ["n_baskets", "PRODUCT_ID"],
        ascending=[False, True],
    )
    .reset_index(drop=True)
)
global_ranked["global_rank"] = global_ranked.index + 1

probe = (
    positives
    .merge(
        personal_ranked[
            ["household_key", "PRODUCT_ID", "personal_rank"]
        ],
        on=["household_key", "PRODUCT_ID"],
        how="left",
        validate="one_to_one",
    )
    .merge(
        global_ranked[["PRODUCT_ID", "global_rank"]],
        on="PRODUCT_ID",
        how="left",
        validate="many_to_one",
    )
)

results = []

for personal_limit, global_limit in [
    (50, 100),
    (200, 100),
    (50, 500),
    (200, 500),
]:
    covered = (
        probe["personal_rank"].le(personal_limit)
        | probe["global_rank"].le(global_limit)
    )
    coverage_by_household = covered.groupby(probe["household_key"])

    results.append({
        "personal_limit": personal_limit,
        "global_limit": global_limit,
        "pair_coverage": covered.mean(),
        "household_mean_coverage": coverage_by_household.mean().mean(),
        "households_without_hits": (~coverage_by_household.any()).sum(),
    })

display(pd.DataFrame(results))
```

```python
print(
    "Доля будущих пар, встречавшихся в личной истории за 180 дней:",
    f"{probe['personal_rank'].notna().mean():.1%}",
)

past_pairs = history[
    ["household_key", "PRODUCT_ID"]
].drop_duplicates()

repeat_check = positives.merge(
    past_pairs,
    on=["household_key", "PRODUCT_ID"],
    how="left",
    indicator=True,
    validate="one_to_one",
)

print(
    "Доля будущих пар, встречавшихся во всей личной истории:",
    f"{repeat_check['_merge'].eq('both').mean():.1%}",
)
```

```python
personal_candidates = personal_ranked.loc[
    personal_ranked["personal_rank"].le(200),
    ["household_key", "PRODUCT_ID"],
]

popular_products = global_ranked.loc[
    global_ranked["global_rank"].le(500),
    ["PRODUCT_ID"],
]

popular_candidates = pd.DataFrame({
    "household_key": eligible_households
}).merge(popular_products, how="cross")

candidates = (
    pd.concat(
        [personal_candidates, popular_candidates],
        ignore_index=True,
    )
    .drop_duplicates(["household_key", "PRODUCT_ID"])
    .merge(
        positives,
        on=["household_key", "PRODUCT_ID"],
        how="left",
        validate="one_to_one",
    )
)

candidates["label"] = candidates["label"].fillna(0).astype("int8")
candidates["cutoff"] = t

print("Пар:", len(candidates))
print("Положительных:", candidates["label"].sum())
print(
    "Размер таблицы, МБ:",
    round(candidates.memory_usage(deep=True).sum() / 1024**2, 1),
)

# Лучший возможный Recall@10 внутри этого списка кандидатов.
future_counts = positives.groupby("household_key").size()
covered_counts = (
    candidates.loc[candidates["label"].eq(1)]
    .groupby("household_key")
    .size()
    .reindex(future_counts.index, fill_value=0)
)

oracle_recall_10 = (
    covered_counts.clip(upper=10) / future_counts
).mean()

print("Верхняя граница среднего Recall@10:", f"{oracle_recall_10:.1%}")
```

```python
import numpy as np

user_item_stats = (
    history.loc[history["DAY"].ge(t - 180)]
    .groupby(["household_key", "PRODUCT_ID"])
    .agg(
        user_baskets_180d=("BASKET_ID", "nunique"),
        last_purchase_day=("DAY", "max"),
    )
    .reset_index()
)

baseline = (
    candidates
    .merge(
        user_item_stats,
        on=["household_key", "PRODUCT_ID"],
        how="left",
        validate="one_to_one",
    )
    .merge(
        global_ranked[["PRODUCT_ID", "n_baskets"]]
        .rename(columns={"n_baskets": "global_baskets_28d"}),
        on="PRODUCT_ID",
        how="left",
        validate="many_to_one",
    )
)

baseline["user_baskets_180d"] = (
    baseline["user_baskets_180d"].fillna(0)
)
baseline["global_baskets_28d"] = (
    baseline["global_baskets_28d"].fillna(0)
)

days_since_purchase = (
    t - baseline["last_purchase_day"]
).fillna(np.inf)

baseline["frequency_recency_score"] = (
    baseline["user_baskets_180d"]
    * 2.0 ** (-days_since_purchase / 30.0)
)
```

```python
def evaluate_ranking(frame, truth, score_col, k=10):
    # Сейчас функция оценивает одну временную точку.
    assert frame["cutoff"].nunique() == 1

    total = truth.groupby("household_key").size()

    sort_cols = ["household_key", score_col]
    ascending = [True, False]

    if score_col != "global_baskets_28d":
        sort_cols.append("global_baskets_28d")
        ascending.append(False)

    sort_cols.append("PRODUCT_ID")
    ascending.append(True)

    top = (
        frame.sort_values(sort_cols, ascending=ascending)
        .groupby("household_key", sort=False)
        .head(k)
        .copy()
    )
    top["position"] = top.groupby("household_key").cumcount() + 1

    hits = (
        top.groupby("household_key")["label"].sum()
        .reindex(total.index, fill_value=0)
    )

    top["gain"] = top["label"] / np.log2(top["position"] + 1)
    dcg = (
        top.groupby("household_key")["gain"].sum()
        .reindex(total.index, fill_value=0)
    )

    ideal = np.concatenate((
        [0.0],
        np.cumsum(1.0 / np.log2(np.arange(1, k + 1) + 1)),
    ))
    idcg = ideal[total.clip(upper=k).to_numpy(dtype=int)]

    return {
        f"Recall@{k}": float((hits / total).mean()),
        f"NDCG@{k}": float((dcg / idcg).mean()),
    }
```

```python
scores = {
    "Популярность": "global_baskets_28d",
    "Личная частота": "user_baskets_180d",
    "Частота + давность": "frequency_recency_score",
}

results = pd.DataFrame([
    {
        "method": name,
        **evaluate_ranking(baseline, positives, score_col),
    }
    for name, score_col in scores.items()
])

display(results)
```

```python
def build_manual_features(purchases, products, candidates, cutoff):
    assert candidates["cutoff"].eq(cutoff).all()

    history = purchases.loc[purchases["DAY"] < cutoff]

    category_cols = [
        "DEPARTMENT",
        "COMMODITY_DESC",
        "SUB_COMMODITY_DESC",
        "BRAND",
    ]

    result = candidates.merge(
        products[["PRODUCT_ID", *category_cols]],
        on="PRODUCT_ID",
        how="left",
        validate="many_to_one",
    )

    for window in [7, 28, 90, 180]:
        recent = history.loc[history["DAY"].ge(cutoff - window)]

        specifications = [
            (
                ["household_key"],
                {
                    f"user_baskets_{window}d": ("BASKET_ID", "nunique"),
                    f"user_products_{window}d": ("PRODUCT_ID", "nunique"),
                },
            ),
            (
                ["PRODUCT_ID"],
                {
                    f"item_baskets_{window}d": ("BASKET_ID", "nunique"),
                    f"item_households_{window}d": ("household_key", "nunique"),
                },
            ),
            (
                ["household_key", "PRODUCT_ID"],
                {
                    f"user_item_baskets_{window}d":
                        ("BASKET_ID", "nunique"),
                },
            ),
        ]

        for keys, aggregations in specifications:
            stats = (
                recent.groupby(keys)
                .agg(**aggregations)
                .reset_index()
            )

            result = result.merge(
                stats,
                on=keys,
                how="left",
                validate="many_to_one",
            )

            count_cols = list(aggregations)
            result[count_cols] = (
                result[count_cols].fillna(0).astype("int32")
            )

        # Доля корзин домохозяйства, содержащих данный товар.
        denominator = result[f"user_baskets_{window}d"].replace(0, np.nan)
        result[f"user_item_basket_share_{window}d"] = (
            result[f"user_item_baskets_{window}d"] / denominator
        ).fillna(0).astype("float32")

    # Интерес к категории товара за 180 дней.
    category_stats = (
        history.loc[history["DAY"].ge(cutoff - 180)]
        .groupby(["household_key", "COMMODITY_DESC"])
        .agg(user_category_baskets_180d=("BASKET_ID", "nunique"))
        .reset_index()
    )

    result = result.merge(
        category_stats,
        on=["household_key", "COMMODITY_DESC"],
        how="left",
        validate="many_to_one",
    )
    result["user_category_baskets_180d"] = (
        result["user_category_baskets_180d"].fillna(0).astype("int32")
    )

    # Давность по всей доступной прошлой истории.
    last_purchase = (
        history.groupby(["household_key", "PRODUCT_ID"])["DAY"]
        .max()
        .rename("last_purchase_day")
        .reset_index()
    )

    result = result.merge(
        last_purchase,
        on=["household_key", "PRODUCT_ID"],
        how="left",
        validate="many_to_one",
    )

    result["days_since_item_purchase"] = (
        cutoff - result["last_purchase_day"]
    ).astype("float32")

    result["item_seen_before"] = (
        result["last_purchase_day"].notna().astype("int8")
    )
    result = result.drop(columns="last_purchase_day")

    result[category_cols] = result[category_cols].astype("category")

    return result, category_cols
```

```python
train_features, categorical_features = build_manual_features(
    purchases, products, candidates, cutoff=t
)

excluded_cols = {"household_key", "PRODUCT_ID", "cutoff", "label"}
feature_cols = [
    col for col in train_features.columns
    if col not in excluded_cols
]

print("Строк:", len(train_features))
print("Признаков:", len(feature_cols))
print(
    "Размер, МБ:",
    round(train_features.memory_usage(deep=True).sum() / 1024**2, 1),
)

display(train_features[
    [
        "user_item_baskets_28d",
        "user_item_basket_share_28d",
        "user_category_baskets_180d",
        "days_since_item_purchase",
        "item_seen_before",
    ]
].head())
```

```python
def make_candidates(
    purchases, cutoff, horizon=7,
    personal_limit=200, popular_limit=500,
):
    assert cutoff + horizon - 1 <= purchases["DAY"].max()

    past = purchases.loc[purchases["DAY"] < cutoff]
    households = past.loc[
        past["DAY"].ge(cutoff - 90), "household_key"
    ].unique()

    personal = (
        past.loc[
            past["DAY"].ge(cutoff - 180)
            & past["household_key"].isin(households)
        ]
        .groupby(["household_key", "PRODUCT_ID"])["BASKET_ID"]
        .nunique()
        .rename("n_baskets")
        .reset_index()
        .sort_values(
            ["household_key", "n_baskets", "PRODUCT_ID"],
            ascending=[True, False, True],
        )
        .groupby("household_key", sort=False)
        .head(personal_limit)
        [["household_key", "PRODUCT_ID"]]
    )

    popular = (
        past.loc[past["DAY"].ge(cutoff - 28)]
        .groupby("PRODUCT_ID")["BASKET_ID"]
        .nunique()
        .rename("n_baskets")
        .reset_index()
        .sort_values(
            ["n_baskets", "PRODUCT_ID"], ascending=[False, True]
        )
        .head(popular_limit)
        [["PRODUCT_ID"]]
    )

    general = pd.DataFrame({
        "household_key": households
    }).merge(popular, how="cross")

    pairs = pd.concat(
        [personal, general], ignore_index=True
    ).drop_duplicates(["household_key", "PRODUCT_ID"])

    truth = (
        purchases.loc[
            purchases["DAY"].ge(cutoff)
            & purchases["DAY"].lt(cutoff + horizon)
            & purchases["household_key"].isin(households),
            ["household_key", "PRODUCT_ID"],
        ]
        .drop_duplicates()
        .assign(label=1)
    )

    pairs = pairs.merge(
        truth,
        on=["household_key", "PRODUCT_ID"],
        how="left",
        validate="one_to_one",
    )
    pairs["label"] = pairs["label"].fillna(0).astype("int8")
    pairs["cutoff"] = cutoff

    return pairs, truth
```

```python
validation_candidates, validation_positives = make_candidates(
    purchases, cutoff=591
)

validation_features, _ = build_manual_features(
    purchases, products, validation_candidates, cutoff=591
)

print("Валидационных пар:", len(validation_features))
print("Положительных:", validation_features["label"].sum())
```

```python
from catboost import CatBoostClassifier

positive_rows = train_features.loc[train_features["label"].eq(1)]
negative_rows = train_features.loc[train_features["label"].eq(0)]

sampled_negatives = negative_rows.sample(
    frac=0.10, random_state=42
)
negative_rate = len(sampled_negatives) / len(negative_rows)

fit_sample = (
    pd.concat([positive_rows, sampled_negatives])
    .sample(frac=1, random_state=42)
    .reset_index(drop=True)
)

weights = np.where(
    fit_sample["label"].eq(1), 1.0, 1.0 / negative_rate
)

categorical_indices = [
    feature_cols.index(col) for col in categorical_features
]

model = CatBoostClassifier(
    iterations=400,
    depth=6,
    learning_rate=0.05,
    l2_leaf_reg=5,
    loss_function="Logloss",
    eval_metric="Logloss",
    random_seed=42,
    thread_count=2,
    task_type="CPU",
    allow_writing_files=False,
)

model.fit(
    fit_sample[feature_cols],
    fit_sample["label"],
    cat_features=categorical_indices,
    sample_weight=weights,
    eval_set=(
        validation_features[feature_cols],
        validation_features["label"],
    ),
    early_stopping_rounds=50,
    use_best_model=True,
    verbose=50,
)
```

```python
validation_scores = validation_features[
    ["household_key", "PRODUCT_ID", "cutoff", "label"]
].copy()

validation_scores["global_baskets_28d"] = (
    validation_features["item_baskets_28d"]
)
validation_scores["personal_frequency"] = (
    validation_features["user_item_baskets_180d"]
)
validation_scores["personal_recency"] = (
    validation_features["user_item_baskets_180d"]
    * 2.0 ** (
        -validation_features["days_since_item_purchase"]
        .fillna(np.inf) / 30.0
    )
)
validation_scores["catboost_score"] = model.predict_proba(
    validation_features[feature_cols]
)[:, 1]

methods = {
    "Популярность": "global_baskets_28d",
    "Личная частота": "personal_frequency",
    "Частота + давность": "personal_recency",
    "CatBoost": "catboost_score",
}

validation_results = pd.DataFrame([
    {
        "method": name,
        **evaluate_ranking(
            validation_scores, validation_positives, score_col
        ),
    }
    for name, score_col in methods.items()
])

display(validation_results)
print("Лучшая итерация, индекс с нуля:", model.get_best_iteration())
```
